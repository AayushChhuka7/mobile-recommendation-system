```python
import pandas as pd
import numpy as np
import re
from rapidfuzz import fuzz, process

df   = GSMArena_All_dataset
df   = df.reset_index(drop=True)
src2 = pd.read_csv('GSMArenaDataset/antutu_score_new.csv')
src3 = pd.read_csv('GSMArenaDataset/Rate_on_antutu_score.csv')
src1 = pd.read_csv('mobiles_data_with_predicted_antutu.csv')

def clean_name(s):
    s = str(s).lower()
    s = re.sub(r"\(\s*\d+\s*gb\s*[+/]\s*\d+\s*gb\s*\)", "", s)
    s = re.sub(r"\b\d+\s*gb\s*[+/]\s*\d+\s*gb\b", "", s)
    s = re.sub(r"\b\d+\s*gb\b", "", s)
    s = re.sub(r"\b(5g|4g|lte|dual sim|global|international)\b", "", s)
    s = re.sub(r"[^a-z0-9\s]", " ", s)
    s = re.sub(r"\s+", " ", s).strip()
    return s

def clean_chipset(s):
    s = str(s).lower()
    s = re.sub(r"\(\d+\s*nm\)", "", s)
    s = re.sub(r"\b(qualcomm|mediatek|samsung|unisoc|apple)\b", "", s)
    s = re.sub(r"\bbionic\b", "", s)
    s = re.sub(r"\b(sm|sdm|msm|apq|qsd|mdm|mt)\d[a-z0-9]*\b", "", s)
    s = re.sub(r"\b(5g|4g)\b", "", s)
    s = re.sub(r"[^a-z0-9\s]", " ", s)
    s = re.sub(r"\s+", " ", s).strip()
    return s

def extract_numbers(s):
    return set(re.findall(r"\d+", s))

def fuzzy_match(key, choices, threshold=90, check_numbers=True):
    if not key:
        return None
    if key in choices:
        return key
    result = process.extractOne(key, choices, scorer=fuzz.token_sort_ratio)
    if not result:
        return None
    match, score, _ = result
    if score < threshold:
        return None
    if check_numbers:
        k_nums, m_nums = extract_numbers(key), extract_numbers(match)
        if k_nums and m_nums and k_nums != m_nums:
            return None
    return match

df['_model_clean']   = df['Model_Name'].apply(clean_name)
df['_chipset_clean'] = df['Chipset'].apply(clean_chipset)

fill_cols = ['AnTuTu_Score', 'GeekBench_Score', 'GPU', 'Chipset', 'Price_INR',
             'Main_Camera_MP', 'Selfie_Camera_MP', 'Storage_GB', 'RAM_GB',
             'Lens_Count', 'Battery_mAh', 'Display_Type']
for col in fill_cols:
    df[f'{col}_Source'] = np.where(df[col].notna(), 'original', 'unfilled')

src2['_phone_clean'] = src2['phone'].apply(clean_name)
src2['_proc_clean']  = src2['processor'].apply(clean_chipset)
p2_name_antutu = src2.dropna(subset=['antutu_v11']).groupby('_phone_clean')['antutu_v11'].median().to_dict()
p2_chip_antutu = src2.dropna(subset=['antutu_v11']).groupby('_proc_clean')['antutu_v11'].median().to_dict()
p2_chip_gpu = (
    src2.dropna(subset=['gpu'])
        .groupby('_proc_clean')['gpu']
        .agg(lambda x: x.mode().iloc[0] if not x.mode().empty else None)
        .to_dict()
)

src3['_key_short'] = src3['Processor_name'].apply(clean_chipset)
src3['_key_full']  = src3['Processor_full_name'].apply(clean_chipset)
p3_lookup = {}
for _, row in src3.iterrows():
    for key in (row['_key_short'], row['_key_full']):
        if key and key not in p3_lookup:
            p3_lookup[key] = row

src1_real = src1[src1['antutu_score_was_predicted'] == False].copy()
src1_real['_model_clean'] = src1_real['Model_Name'].apply(clean_name)
src1_real['_chipset_clean'] = src1_real['Chipset'].apply(clean_chipset)
src1_real = src1_real.drop_duplicates(subset='_model_clean')
p1_name_antutu = src1_real.set_index('_model_clean')['antutu_score_final'].to_dict()
p1_chip_antutu = src1_real.dropna(subset=['antutu_score_final']).groupby('_chipset_clean')['antutu_score_final'].median().to_dict()

src1_all = src1.copy()
src1_all['_model_clean'] = src1_all['Model_Name'].apply(clean_name)
src1_all = src1_all.drop_duplicates(subset='_model_clean')
src1_all_by_name = src1_all.set_index('_model_clean')
p1_all_keys = list(src1_all_by_name.index)

mask = df['AnTuTu_Score'].isna()
keys = list(p2_name_antutu.keys())
for idx in df.index[mask]:
    m = fuzzy_match(df.at[idx, '_model_clean'], keys, threshold=90, check_numbers=False)
    if m:
        df.at[idx, 'AnTuTu_Score'] = p2_name_antutu[m]
        df.at[idx, 'AnTuTu_Score_Source'] = 'src2_name'

mask = df['AnTuTu_Score'].isna()
keys = list(p2_chip_antutu.keys())
for idx in df.index[mask]:
    m = fuzzy_match(df.at[idx, '_chipset_clean'], keys, threshold=88, check_numbers=True)
    if m:
        df.at[idx, 'AnTuTu_Score'] = p2_chip_antutu[m]
        df.at[idx, 'AnTuTu_Score_Source'] = 'src2_chipset'

mask = df['AnTuTu_Score'].isna()
keys = [k for k in p3_lookup if pd.notna(p3_lookup[k]['antutu_v11'])]
for idx in df.index[mask]:
    m = fuzzy_match(df.at[idx, '_chipset_clean'], keys, threshold=88, check_numbers=True)
    if m:
        df.at[idx, 'AnTuTu_Score'] = p3_lookup[m]['antutu_v11']
        df.at[idx, 'AnTuTu_Score_Source'] = 'src3_chipset'

mask = df['AnTuTu_Score'].isna()
keys = list(p1_name_antutu.keys())
for idx in df.index[mask]:
    m = fuzzy_match(df.at[idx, '_model_clean'], keys, threshold=90, check_numbers=False)
    if m:
        df.at[idx, 'AnTuTu_Score'] = p1_name_antutu[m]
        df.at[idx, 'AnTuTu_Score_Source'] = 'src1_name'

mask = df['AnTuTu_Score'].isna()
keys = list(p1_chip_antutu.keys())
for idx in df.index[mask]:
    m = fuzzy_match(df.at[idx, '_chipset_clean'], keys, threshold=88, check_numbers=True)
    if m:
        df.at[idx, 'AnTuTu_Score'] = p1_chip_antutu[m]
        df.at[idx, 'AnTuTu_Score_Source'] = 'src1_chipset'

mask = df['GeekBench_Score'].isna()
keys = [k for k in p3_lookup if pd.notna(p3_lookup[k]['gb6_multi'])]
for idx in df.index[mask]:
    m = fuzzy_match(df.at[idx, '_chipset_clean'], keys, threshold=88, check_numbers=True)
    if m:
        df.at[idx, 'GeekBench_Score'] = p3_lookup[m]['gb6_multi']
        df.at[idx, 'GeekBench_Score_Source'] = 'src3_chipset'

mask = (df['GPU'].isna()) | (df['GPU'].astype(str).str.strip() == '')
keys2 = list(p2_chip_gpu.keys())
keys3 = [k for k in p3_lookup if pd.notna(p3_lookup[k]['gpu'])]
for idx in df.index[mask]:
    m = fuzzy_match(df.at[idx, '_chipset_clean'], keys2, threshold=88, check_numbers=True)
    if m and p2_chip_gpu.get(m):
        df.at[idx, 'GPU'] = p2_chip_gpu[m]
        df.at[idx, 'GPU_Source'] = 'src2_chipset'
        continue
    m3 = fuzzy_match(df.at[idx, '_chipset_clean'], keys3, threshold=88, check_numbers=True)
    if m3:
        df.at[idx, 'GPU'] = p3_lookup[m3]['gpu']
        df.at[idx, 'GPU_Source'] = 'src3_chipset'

mask = df['Chipset'].isna()
for idx in df.index[mask]:
    m = fuzzy_match(df.at[idx, '_model_clean'], p1_all_keys, threshold=90, check_numbers=False)
    if m and pd.notna(src1_all_by_name.at[m, 'Chipset']):
        df.at[idx, 'Chipset'] = src1_all_by_name.at[m, 'Chipset']
        df.at[idx, 'Chipset_Source'] = 'src1_name'

spec_map = {
    'Price_INR':        'Price_INR',
    'Main_Camera_MP':   'Main_Camera_MP',
    'Selfie_Camera_MP': 'Selfie_Camera_MP',
    'Storage_GB':       'Storage_GB',
    'RAM_GB':           'Ram_GB',
    'Lens_Count':       'Lens_Count',
    'Battery_mAh':      'Battery_Mah',
    'Display_Type':     'Display_Type',
}
for gsm_col, src_col in spec_map.items():
    mask = df[gsm_col].isna()
    for idx in df.index[mask]:
        m = fuzzy_match(df.at[idx, '_model_clean'], p1_all_keys, threshold=90, check_numbers=False)
        if m and pd.notna(src1_all_by_name.at[m, src_col]):
            df.at[idx, gsm_col] = src1_all_by_name.at[m, src_col]
            df.at[idx, f'{gsm_col}_Source'] = 'src1_name'

df.drop(columns=['_model_clean', '_chipset_clean'], inplace=True)
for col in fill_cols:
    df[f'{col}_is_imputed'] = df[f'{col}_Source'] != 'original'

print(f"AnTuTu_Score      {df['AnTuTu_Score'].notna().mean()*100:.1f}%")
print(f"GeekBench_Score   {df['GeekBench_Score'].notna().mean()*100:.1f}%")
print(f"GPU               {df['GPU'].notna().mean()*100:.1f}%")
print(f"Chipset           {df['Chipset'].notna().mean()*100:.1f}%")
print(f"Price_INR         {df['Price_INR'].notna().mean()*100:.1f}%")
print(f"Main_Camera_MP    {df['Main_Camera_MP'].notna().mean()*100:.1f}%")
print(f"Selfie_Camera_MP  {df['Selfie_Camera_MP'].notna().mean()*100:.1f}%")
print(f"Storage_GB        {df['Storage_GB'].notna().mean()*100:.1f}%")
print(f"RAM_GB            {df['RAM_GB'].notna().mean()*100:.1f}%")
print(f"Lens_Count        {df['Lens_Count'].notna().mean()*100:.1f}%")
print(f"Battery_mAh       {df['Battery_mAh'].notna().mean()*100:.1f}%")
print(f"Display_Type      {df['Display_Type'].notna().mean()*100:.1f}%")

df.to_csv('GSMArena_AnTuTu_Final_two.csv', index=False)
```

```python
canonical_brands = sorted(new_df['Brand'].dropna().unique(), key=len, reverse=True)

def derive_brand(model_name, current_brand):
    if pd.isna(model_name):
        return current_brand
    name_lower = str(model_name).lower()
    for b in canonical_brands:
        if name_lower.startswith(b.lower()):
            return b
    return current_brand  

n_before = new_df['Brand'].copy()
new_df['Brand'] = new_df.apply(lambda r: derive_brand(r['Model_Name'], r['Brand']), axis=1)
print(f"Rows corrected: {(new_df['Brand'] != n_before).sum()}")  
new_df.duplicated().sum()
```

```python
import numpy as np
conditions = [
    new_df["Price_EUR"].isna() & new_df["Price_INR"].notna(),
    new_df["Price_EUR"].isna() & new_df["Price_USD"].notna(),
    new_df["Price_EUR"].isna() & new_df["Price_GBP"].notna(),
]

choices = [
    new_df["Price_INR"] / 109.0,  # INR to EUR
    new_df["Price_USD"] * 0.87,  # USD to EUR
    new_df["Price_GBP"] * 1.16,  # GBP to EUR
]

# 3. Apply the changes. If no condition is met, keep the original Price_EUR
new_df["Price_EUR"] = np.select(conditions, choices, default=new_df["Price_EUR"])
```

```python
def auto_fill_display_size(dim_string):
    if pd.isna(dim_string) or str(dim_string).strip() == '-':
        return np.nan
    try:
        metric_part = str(dim_string).split("(")[0]
        nums = re.findall(r"[-+]?\d*\.\d+|\d+", metric_part)
        if len(nums) < 2:
            return np.nan
        height_mm, width_mm = float(nums[0]), float(nums[1])
        body_diagonal = np.sqrt((height_mm / 25.4) ** 2 + (width_mm / 25.4) ** 2)
        aspect_ratio = height_mm / width_mm if width_mm > 0 else 1

        if aspect_ratio >= 2.5:
            bezel_frac = 0.78        # ultra-narrow "lipstick" devices
        elif body_diagonal < 6:
            bezel_frac = 0.42        # small candybar / flip phones
        elif body_diagonal < 9:
            bezel_frac = 0.11        # standard smartphones
        else:
            bezel_frac = 0.13        # tablets

        screen_diagonal = body_diagonal * (1 - bezel_frac)
        return max(0.5, round(screen_diagonal, 2))
    except Exception:
        return np.nan

mask = new_df['Display_Size_inch'].isna()
new_df.loc[mask, 'Display_Size_inch'] = new_df.loc[mask, 'Dimensions'].apply(auto_fill_display_size)

new_df['Display_Size_inch'].isnull().mean() * 100
```

### CPU Feature Engineering

```python
import re

CORE_WORDS = {
    "single": 1, "dual": 2, "triple": 3, "quad": 4, "penta": 5,
    "hexa": 6, "hepta": 7, "octa": 8, "nona": 9, "deca": 10,
}

def extract_core_count(s):
    """'Octa-core (...)' -> 8, '8-core (1+3+4) @ ...' -> 8,
    '1.0 GHz Cortex-A9' (no 'core' word at all) -> 1 (single core)."""
    if pd.isna(s):
        return None
    s_low = str(s).lower()
    for word, n in CORE_WORDS.items():
        if re.match(rf"^{word}-core", s_low):
            return n
    m = re.match(r"^(\d+)-core", s_low)
    if m:
        return int(m.group(1))
    if re.match(r"^[\d.]+\s*(ghz|mhz)", s_low):
        return 1
    return None

def extract_max_clock_ghz(s):
    """Pull every GHz/MHz value mentioned and return the highest one --
    that's the peak-performance core's speed, the most useful single number
    for comparing chips. Handles mixed-unit strings like '8-core ... @ 2300 MHz'."""
    if pd.isna(s):
        return None
    s = str(s)
    ghz_vals = [float(x) for x in re.findall(r"([\d.]+)\s*GHz", s, re.IGNORECASE)]
    mhz_vals = [float(x) / 1000 for x in re.findall(r"([\d.]+)\s*MHz", s, re.IGNORECASE)]
    all_vals = ghz_vals + mhz_vals
    return round(max(all_vals), 2) if all_vals else None

def extract_architecture(s):
    """Pull out the named core architecture, in priority order (most
    specific/modern first). When a string lists multiple core names (a
    big.LITTLE design naming both the performance and efficiency core),
    this keeps the FIRST/highest-priority one found, which is normally
    the performance core -- the one that matters most for benchmarks."""
    if pd.isna(s):
        return None
    s = str(s)
    patterns = [
        r"(Oryon(?:\s*V\d)?)",
        r"(Cortex-X\d+)",
        r"(Cortex-A\d+)",
        r"(Cortex-\d{3})",
        r"(C1-(?:Ultra|Premium|Pro|Nano))",
        r"(Kryo(?:\s*\d+)?)",
        r"(Krait(?:\s*\d+)?)",
        r"(Scorpion)",
        r"(ARM\s*11)",
        r"(Twister|Zephyr|Hurricane|Monsoon|Mistral|Typhoon|Avalanche|Blizzard)",
    ]
    for pat in patterns:
        m = re.findall(pat, s, re.IGNORECASE)
        if m:
            return m[0]
    return None

new_df["CPU_core_count"]    = new_df["CPU"].apply(extract_core_count)
new_df["CPU_max_clock_ghz"] = new_df["CPU"].apply(extract_max_clock_ghz)
new_df["CPU_architecture"]  = new_df["CPU"].apply(extract_architecture)

n_total = new_df["CPU"].notna().sum()
print(f"Coverage out of {n_total} rows with a CPU value:")
print(f"  core_count:    {new_df['CPU_core_count'].notna().sum()} ({new_df['CPU_core_count'].notna().mean()*100:.1f}%)")
print(f"  max_clock_ghz: {new_df['CPU_max_clock_ghz'].notna().sum()} ({new_df['CPU_max_clock_ghz'].notna().mean()*100:.1f}%)")
print(f"  architecture:  {new_df['CPU_architecture'].notna().sum()} ({new_df['CPU_architecture'].notna().mean()*100:.1f}%)")
```

```python
import re
import pandas as pd
import numpy as np

def clean_and_engineer_chipset(df):
    # Ensure raw column is treated as string and lowercase for uniform matching
    df['Chipset'] = df['Chipset'].fillna('Unknown').astype(str)
    
    # -------------------------------------------------------------
    # 1. Process Node (Chipset_nm) -> Numeric
    # -------------------------------------------------------------
    df['Chipset_nm'] = df['Chipset'].str.extract(r'\((\d+)\s*nm\)').astype(float)
    
    # -------------------------------------------------------------
    # 2. Chipset_Brand & Chipset_Family
    # -------------------------------------------------------------
    def get_brand_and_family(text):
        text_lower = text.lower()
        if 'qualcomm' in text_lower or 'snapdragon' in text_lower:
            return 'Qualcomm', 'Snapdragon'
        elif 'mediatek' in text_lower or 'helio' in text_lower or 'dimensity' in text_lower:
            brand = 'Mediatek'
            family = 'Dimensity' if 'dimensity' in text_lower else 'Helio'
            return brand, family
        elif 'apple' in text_lower:
            return 'Apple', 'A-Series' if ' bionic' in text_lower or ' a1' in text_lower else 'M-Series'
        elif 'exynos' in text_lower:
            return 'Samsung', 'Exynos'
        elif 'google' in text_lower or 'tensor' in text_lower:
            return 'Google', 'Tensor'
        elif 'kirin' in text_lower:
            return 'Huawei', 'Kirin'
        elif 'unisoc' in text_lower:
            return 'Unisoc', 'T-Series'

        elif 'spreadtrum' in text_lower:
            return 'Spreadtrum', 'SC-Series'

        elif 'intel atom' in text_lower or 'intel' in text_lower:
            return 'Intel', 'Atom'
        
        elif 'tegra' in text_lower or 'nvidia' in text_lower:
            return 'Nvidia', 'Tegra'
        
        elif 'omap' in text_lower or 'ti omap' in text_lower:
            return 'Texas_Instruments', 'OMAP'
        
        elif 'rockchip' in text_lower:
            return 'Rockchip', 'RK-Series'
        
        elif 'k3v' in text_lower:
            return 'Huawei', 'K3V'
            
        elif 'marvell' in text_lower:
            return 'Marvell', 'PXA'

        elif 'broadcom' in text_lower:
            return 'Broadcom', 'BCM'
        
        elif 'novathor' in text_lower:
            return 'ST-Ericsson', 'NovaThor'
        else:
            return 'Unknown', 'Unknown'
            
    brands_families = df['Chipset'].apply(get_brand_and_family)
    df['Chipset_Brand'] = [b[0] for b in brands_families]
    df['Chipset_Family'] = [b[1] for b in brands_families]

    # -------------------------------------------------------------
    # 3. Chipset_Generation
    # -------------------------------------------------------------
    def extract_generation(text):

        text = text.lower()
 
        if 'elite' in text:
            return 'SD_Elite'

        m = re.search(r'gen\s*(\d+)', text)
        if m:
            return f"SD_Gen{m.group(1)}"

        m = re.search(r'snapdragon\s*s(\d)', text)
        if m:
            return f"SD_S{m.group(1)}"
    
        # Helio G
        m = re.search(r'helio\s*g(\d+)', text)
        if m:
            return f"Helio_G{m.group(1)}"

        m = re.search(r'helio\s*p(\d+)', text)
        if m:
            return f"Helio_P{m.group(1)}"

        m = re.search(r'helio\s*a(\d+)', text)
        if m:
            return f"Helio_A{m.group(1)}"

        m = re.search(r'dimensity\s*(\d+)', text)
        if m:
            return f"Dimensity_{m.group(1)}"

        m = re.search(r'kirin\s*(\d+)', text)
        if m:
            return f"Kirin_{m.group(1)}"

        m = re.search(r'tensor\s*g(\d+)', text)
        if m:
            return f"Tensor_G{m.group(1)}"
    
        m = re.search(r'apple\s*a(\d+)', text)
        if m:
            return f"A{m.group(1)}"

        m = re.search(r'\bm(\d+)\b', text)
        if m and 'apple' in text:
            return f"M{m.group(1)}"
    
        m = re.search(r'exynos\s*(\d+)', text)
        if m:
            return f"Exynos_{m.group(1)}"
    
        m = re.search(r'(?:tiger\s*)?t(\d+)', text)
        if m:
            return f"T{m.group(1)}"

        m = re.search(r'snapdragon\s*(\d{3})', text)
        if m:
            return f"Snapdragon_{m.group(1)}"
        
        m = re.search(r'snapdragon\s*(\d{2})', text)
        if m:
            return f"Snapdragon_{m.group(1)}"
    
        return 'Legacy'

    df['Chipset_Generation'] = df['Chipset'].apply(extract_generation)


    flagship_series = ['elite', 'ultra', 'pro', 'extreme', 'ultimate', '9400', '9500', '9300', 'm4']
    df['Chipset_Is_Flagship'] = df.apply(
        lambda row: 1 if any(word in row['Chipset'].lower() for word in flagship_series) 
                         or row['Chipset_nm'] <= 3.0 
                         or 'snapdragon 8' in row['Chipset'].lower() else 0, axis=1
    )


    df['Chipset_Has_5G_Tag'] = df['Chipset'].apply(lambda x: 1 if '5g' in x.lower() else 0)
    df['Has_Pro']            = df['Chipset'].apply(lambda x: 1 if 'pro' in x.lower() else 0)
    df['Has_Ultra']          = df['Chipset'].apply(lambda x: 1 if 'ultra' in x.lower() else 0)
    df['Has_Plus']           = df['Chipset'].apply(lambda x: 1 if '+' in x or 'plus' in x.lower() else 0)
    df['Has_Elite']          = df['Chipset'].apply(lambda x: 1 if 'elite' in x.lower() else 0)

    return df

new_df = clean_and_engineer_chipset(new_df)
```

### FIll Camera_Features with Featre Engineering

```python
print("Initial:", new_df['Camera_Features'].isna().sum())  
new_df['Camera_Features'] = new_df['Camera_Features'].fillna('Unknown')
print("After Step :", new_df['Camera_Features'].isna().sum())
```

```python
import pandas as pd
import numpy as np
import re

def engineer_camera_features(df):
    df['Camera_Features'] = df['Camera_Features'].fillna('Unknown').astype(str)

    df['Has_HDR'] = df['Camera_Features'].str.contains(
        r'\bHDR\b|Ultra HDR|Auto-HDR|auto HDR',
        case=False,
        regex=True
    ).astype(int)

    df['Has_Panorama'] = df['Camera_Features'].str.contains('panorama', case=False).astype(int)
    df['Has_Color_Spectrum'] = df['Camera_Features'].str.contains('color spectrum', case=False).astype(int)
    df['Has_Pixel_Shift'] = df['Camera_Features'].str.contains('pixel shift', case=False).astype(int)
    df['Has_Best_Face'] = df['Camera_Features'].str.contains('best face', case=False).astype(int)
    df['Has_Ultra_HDR'] = df['Camera_Features'].str.contains('ultra hdr', case=False).astype(int)

    def flash_type(text):
        text = text.lower()
        if 'xenon' in text: return 'Xenon'
        elif 'hexa-led' in text: return 'Hexa_LED'
        elif 'penta-led' in text: return 'Penta_LED'
        elif 'quad-led' in text: return 'Quad_LED'
        elif 'triple-led' in text: return 'Triple_LED'
        elif 'dual-led' in text or 'dual led' in text: return 'Dual_LED'
        elif 'ring-led' in text or 'ring led' in text: return 'Ring_LED'
        elif 'led flash' in text: return 'LED'
        else: return 'None'

    df['Flash_Type'] = df['Camera_Features'].apply(flash_type)

    def lens_brand(text):
        text = text.lower()
        if 'leica' in text: return 'Leica'
        elif 'zeiss' in text: return 'Zeiss'
        elif 'hasselblad' in text: return 'Hasselblad'
        elif 'ricoh' in text: return 'Ricoh'
        elif 'schneider' in text: return 'Schneider'
        else: return 'None'

    df['Lens_Brand'] = df['Camera_Features'].apply(lens_brand)

    def af_type(text):
        text = text.lower()
        if 'multi-zone laser af' in text: return 'Multi_Zone'
        elif 'single-zone laser af' in text: return 'Single_Zone'
        elif 'lidar af' in text: return 'LiDAR'
        elif 'laser af' in text or 'laser focus' in text: return 'Laser_AF'
        else: return 'None'

    df['AF_Type'] = df['Camera_Features'].apply(af_type)

    return df 

new_df = engineer_camera_features(new_df) 
new_df.drop(columns=['Camera_Features'], inplace=True)
```

### GPU Featre Engineering and Fillna

```python
print(new_df['GPU'].isna().sum())
# 1
new_df['GPU'] = (new_df.groupby('Chipset' ,dropna=False)['GPU'].transform(lambda x: x.fillna(x.mode().iloc[0]) if not x.mode().empty else x))
print(new_df['GPU'].isna().sum())

new_df['GPU'] = (new_df.groupby('CPU' ,dropna=False)['GPU'].transform(lambda x: x.fillna(x.mode().iloc[0]) if not x.mode().empty else x))
print(new_df['GPU'].isna().sum()) 

new_df['GPU'] = (new_df.groupby(['Chipset_Brand', 'Chipset_Family'] ,dropna=False)['GPU'].transform(lambda x: x.fillna(x.mode().iloc[0]) if not x.mode().empty else x))
print(new_df['GPU'].isna().sum()) 

new_df['GPU'] = (new_df.groupby(['Brand', 'Announced_Year'], dropna=False)['GPU'].transform(lambda x: x.fillna(x.mode().iloc[0]) if not x.mode().empty else x))
print(new_df['GPU'].isna().sum()) 

new_df['GPU'] = new_df['GPU'].fillna('Unknown')
print(new_df['GPU'].isna().sum())
```

```python
import pandas as pd
import re

def engineer_gpu_features(df):
    df['GPU'] = df['GPU'].fillna('Unknown').astype(str)

    # -------------------------------------------------
    # 1. GPU Brand
    # -------------------------------------------------
    def gpu_brand(text):
        text = text.lower()

        if 'adreno' in text:
            return 'Adreno'
        elif 'immortalis' in text:
            return 'Immortalis'
        elif 'mali' in text:
            return 'Mali'
        elif 'apple gpu' in text:
            return 'Apple'
        elif 'powervr' in text:
            return 'PowerVR'
        elif 'img' in text:
            return 'IMG'
        elif 'maleoon' in text:
            return 'Maleoon'
        elif 'xclipse' in text:
            return 'Xclipse'
        elif 'vivante' in text:
            return 'Vivante'
        elif 'intel' in text:
            return 'Intel'
        elif 'geforce' in text:
            return 'GeForce'
        else:
            return 'Other'

    df['GPU_Brand'] = df['GPU'].apply(gpu_brand)

    # -------------------------------------------------
    # 2. GPU Family
    # -------------------------------------------------
    def gpu_family(text):
        text = text.lower()

        if 'adreno' in text:
            return 'Adreno'

        elif 'immortalis' in text:
            return 'Immortalis'

        elif 'mali-g' in text:
            return 'Mali_G'

        elif 'mali-t' in text:
            return 'Mali_T'

        elif 'mali-400' in text or 'mali-450' in text:
            return 'Mali_400'

        elif 'powervr ge' in text:
            return 'PowerVR_GE'

        elif 'powervr sgx' in text:
            return 'PowerVR_SGX'

        elif 'powervr gx' in text:
            return 'PowerVR_GX'

        elif 'apple gpu' in text:
            return 'Apple_GPU'

        elif 'xclipse' in text:
            return 'Xclipse'

        elif 'maleoon' in text:
            return 'Maleoon'

        elif 'img' in text:
            return 'IMG'

        elif 'vivante' in text:
            return 'Vivante'

        elif 'geforce' in text:
            return 'GeForce'

        else:
            return 'Other'

    df['GPU_Family'] = df['GPU'].apply(gpu_family)

    # -------------------------------------------------
    # 3. Flagship GPU
    # -------------------------------------------------
    flagship_keywords = [
        'adreno 830','adreno 840','adreno 825',
        'adreno 740','adreno 750','adreno 730',
        'immortalis',
        'g720','g710','g715','g925',
        'apple gpu',
        'xclipse 940','xclipse 960'
    ]

    df['GPU_Is_Flagship'] = (
        df['GPU']
        .str.lower()
        .apply(lambda x: int(any(k in x for k in flagship_keywords)))
    )

    # -------------------------------------------------
    # 4. Multi Variant
    # -------------------------------------------------
    keywords = [
        '/',
        ' or ',
        'usa',
        'china',
        'global',
        'international',
        'wifi',
        'lte',
        'version',
        'model'
    ]

    df['GPU_Is_MultiVariant'] = (
        df['GPU']
        .str.lower()
        .apply(lambda x: int(any(k in x for k in keywords)))
    )

    return df

new_df = engineer_gpu_features(new_df)
new_df.drop(columns=['GPU_Is_MultiVariant'], inplace=True)
```
